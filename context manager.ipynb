```python
class Emp:
    def __init__(self,name):
        self.name = name
    def __enter__(self):
        print("this is enter method")
    def __exit__(self,*args):
        print("this is exit method")
with Emp("asee") as e:
    print(e.name)
```

```python
class Emp:
    def __init__(self,name):
        self.name = name
    def __enter__(self):
        print("this is enter method")
        return self
    def __exit__(self,*args):
        print("this is exit method")
with Emp("asee") as e:
    print(e.name)
```
