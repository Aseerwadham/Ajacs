```python
#3. Write a code to get the output dictionary {70: ['rahul', 'kohli'], 50: ['jadeja', 'panth'], 40: ['rakesh']}
#input{"rahul":70,"jadeja":50,"kohli":70,"panth":50,"rakesh":40}

l={"rahul":70,"jadeja":50,"kohli":70,"panth":50,"rakesh":40}
res={}
for key,value in l.items():
    if value in res:
        res[value].append(key)
    else:
        res.update({value: [key]})
print(res)
```

```python
##4. Write a python program that:
# takes a string of comma-separated words as input. converts all words that have more than 5 characters to uppercae
#and those with 5 or fewer characters to lowercase. Then concentrate all the words back together, separated by hyphens(-).
# output the modified string.

str = "asee, aseerwadham, sambasiva"
str1 = str.split(",")
res = [ ]
for i in str1:
    if len(i) > 5:
        i = i.upper()
    else:
        i = i.lower()    
    res.append(i)
print("-".join(res))        
```

```python
# 5. Merge the dictionaries for each common key is the sum of values of both dictionaries.
A = {'a': 5, 'b': 10, 'c': 15}
B = {'b': 3, 'c': 7, 'd': 20}
merged = {}
for key in A:
    merged[key] = A[key]
for key in B:
    if key in merged:
        merged[key] += B[key]
    else:
        merged[key] = B[key]
print(merged)
```

```python
# 6. Write a python program that:
#accepts a list of integers and an integer target value as input.
#Finds all unique pairs of integers from the list that sum up to the given target value.
#Returns a list of pairs as tuples, where each tuple consists of two integers, and the pair is in ascending order.
#Ensure that no pair appears twice in the result.
#Ex - [1,2,3,4,5,6,7,8], target value: 10
#output : [(2,8),(3,7),(4,6)]


lst = [1,2,3,4,5,6,7,8]
K = 10
result = []
for i in range(len(lst)):
    for j in range(i + 1, len(lst)):
        if lst[i] + lst[j] == K:
            result.append((lst[i], lst[j]))
print(result)
```
