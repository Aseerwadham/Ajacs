```python
#string.rstrip([chars]); rstrip() only removes characters from the end (right side) of the string.
#If chars is provided, it removes any combination of the characters listed until it encounters a character not in the set.
#If the string does not end with the specified characters, the original string is returned unchanged.
s = "Hello, world!   "
result = s.rstrip()
print(f"'{result}'") 
s = "Hello, world!!!"
result = s.rstrip('!')
print(result) 
s = "123Hello456"
result = s.rstrip('456')
print(result) 
s.rstrip?
```

```python
#string.split(sep, maxsplit); split() returns a list of substrings.
#If sep is not specified, it splits on any whitespace and removes extra whitespace.
#If maxsplit is specified, it limits the number of splits performed. The last element of the returned list will include the remainder of the string.
s = "apple        banana cherry"
result = s.split()
print(result) 
s = "apple,banana,cherry"
result = s.split(",")
print(result)  
s = "apple banana cherry"
result = s.split(" ", 1)  # Only split once
print(result)  
s = "apple,banana,cherry,"
result = s.split(",")
print(result)  
s.split?
```

```python
#string.splitlines(keepends=False); splitlines() is useful for processing multiline strings.
#It handles different types of line breaks, making it versatile for different operating systems.
#The method returns a list of lines, and you can choose whether or not to keep the line breaks by using the keepends parameter.
s = "Hello\nWorld\nThis is Python."
result = s.splitlines()
print(result)  
s = "Hello\nWorld\nThis is Python."
result = s.splitlines(keepends=True)
print(result)  
s = "Hello\r\nWorld\rThis is Python."
result = s.splitlines()
print(result) 
s.splitlines?
```

```python
#string.startswith(prefix, start, end) ;startswith() is case-sensitive.
#You can use a tuple of prefixes to check against multiple options at once.
#The method is particularly useful for validating input or parsing strings based on known patterns.
s = "Hello, world!"
result = s.startswith("Hello")
print(result) 
s = "Hello, world!"
result = s.startswith("world")
print(result) 
s = "Hello, world!"
result = s.startswith("world", 7)  # Starts checking from index 7
print(result) 
s = "Hello, world!"
result = s.startswith(("Hi", "Hello"))
print(result) 
s.startswith?
```

```python
#string.strip([chars]); strip() only removes characters from the beginning and the end of the string.
#If chars is provided, it removes any combination of the characters listed until it encounters a character not in the set.
#If the string does not start or end with the specified characters, the original string is returned unchanged.
#The method does not modify the original string but returns a new one.
s = "   Hello, world!   "
result = s.strip()
print(f"'{result}'") 
s = "###Hello, world!###"
result = s.strip('#')
print(result) 
s = "abcHello"
result = s.strip('abcxyz')
print(result) 
s.strip?
```

```python
#string.swapcase(); swapcase() does not modify the original string; it returns a new string with the case swapped.
#The method is useful for toggling the case of a string, often for formatting or display purposes.
#It only affects alphabetic characters; numbers and punctuation remain unchanged.
s = "Hello, World!"
result = s.swapcase()
print(result) 
s = "python programming"
result = s.swapcase()
print(result) 
s = "PYTHON PROGRAMMING"
result = s.swapcase()
print(result) 
s.swapcase?
```

```python
#string.title(); title() does not modify the original string; it returns a new string with the title case applied.
#It is not always perfect for contractions or words with special characters (like apostrophes), as shown in the example above.
#It is useful for formatting strings for titles, headings, or names where title case is desired.
s = "hello world"
result = s.title()
print(result)  
s = "python programming is FUN"
result = s.title()
print(result)  
s = "it's a beautiful day"
result = s.title()
print(result)  
s.title?
```

```python
#string.translate(table); translate() does not modify the original string; it returns a new string with the specified character replacements.
#The translation table must be created with maketrans().
#This method is particularly useful for character substitution and removal in strings.

table = str.maketrans("abc", "123")
s = "abcde"
result = s.translate(table)
print(result)  

table = str.maketrans("", "", "aeiou")  # Deletes vowels
s = "Hello, world!"
result = s.translate(table)
print(result)  

table = str.maketrans({"H": "h", "W": "w"})
s = "Hello, World!"
result = s.translate(table)
print(result)  
s.translate?
```

```python
#string.upper(); upper() converts all alphabetic characters in the string to uppercase.
#It does not modify the original string; it returns a new string with the uppercase transformation.
#Non-alphabetic characters (numbers, spaces, punctuation) are not affected.
s = "Hello, World!"
result = s.upper()
print(result) 
s = "python programming"
result = s.upper()
print(result)
s = "Python3.9"
result = s.upper()
print(result) 
s.upper?
```

```python
#string.zfill(width); zfill() pads the string on the left with zeros until the specified width is reached.
#It handles negative numbers correctly by keeping the minus sign at the start.
#If the string's length is already greater than or equal to the specified width, it returns the original string unchanged.
s = "42"
result = s.zfill(5)
print(result)  
s = "12345"
result = s.zfill(3)
print(result) 
s = "-42"
result = s.zfill(5)
print(result) 
s.zfill?
```
