```python
from copy import deepcopy
s1={10,20,30,40}
s2=s1
s3=s1.copy()
s4=deepcopy(s1)
s1.add(100)
print("s1=",s1)
print("s2=",s2)
print("s3=",s3)
print("s4=",s4)
```

```python
s1={1000,(1000,2000,3000),4567.89}
s2=s1.copy()
s3=deepcopy(s1)
for i in s1:
    print(i,id(i))
print("*"*20)
for i in s2:
    print(i, id(i))
print("*"*20)
for i in s3:
    print(i, id(i))
```
