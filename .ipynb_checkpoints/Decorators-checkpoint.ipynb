```python
def add(x,y):
    return x+y
def add_dec(f):
    def inner(x,y):
        print(f"x={x},y={y}")
        r1=f(x,y)
        print("r1=",r1)
    return inner
res=add_dec(add)
r2=res(10,20)
print("r2=",r2)
```

```python
def add_dec(f):
    def inner(x,y):
        print(f"x={x},y={y}")
        r1=f(x,y)
        print("r1=",r1)
    return inner
def add(x,y):
    return x+y
res=add(10,20)
print("res=",res)
```

```python
def add_dec(f):
    def inner(x,y):
        print(f"x={x},y={y}")
        r1=f(x,y)
        print("r1=",r1)
    return inner  ##closure : which should have two functions like outer and inner, also outer function return inner funtion

@add_dec  ## decorator , we can able to write decarators only if there is a closure
def add(x,y):
    return x+y
res=add(10,20) ## res=add_dec(add)(10,20)
print("res=",res)
```
