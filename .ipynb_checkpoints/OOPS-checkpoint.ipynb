```python
#Class has two object types 1. class object 2. Data object
# Emp --> class obj, Emp()--> data obj
# Data obj's having power to access anything in class obj(Emp)
class Emp:
    company="AJA" ## class attributes
jay=Emp()
jay.name="jairam" ## data or instance attributes
jay.id=1 ## data or instance attributes
vedha=Emp()
vedha.name="vedhanvesh"
vedha.id=2
print(Emp.company)
print(Emp.name)
```

```python
class Emp:
    company="AJA" ## class attributes
jay=Emp()
jay.name="jairam" ## data or instance attributes
jay.id=1 ## data or instance attributes
vedha=Emp()
vedha.name="vedhanvesh"
vedha.id=2
print(jay.company)
print(vedha.company)
print(jay.name)
print(vedha.name)
print(jay.id)
print(vedha.id)
```

```python
## class object only access class attributes, but data object can access class attributes also.
```

```python
class Emp:
    company = "AJA"
    def get_details(obj):
        print(obj.name, obj.id)
jay = Emp()
jay.name = "jairam"
jay.id = 1
vedha = Emp()
vedha.name = "vedhanvesh"
vedha.id = 2
Emp.get_details(jay)
Emp.get_details(vedha)        
```

```python
class Emp:
    company = "AJA"
    def get_details(obj):
        print(obj.name, obj.id)
jay = Emp()
jay.name = "jairam"
jay.dob = "28-10-1997" ## extra data arguments even passing not showing any error, but the data is clumzy.
jay.id = 1             # Here we are passing attributes outside the class
vedha = Emp()
vedha.name = "vedhanvesh"
vedha.id = 2
Emp.get_details(jay)
Emp.get_details(vedha)        
```

```python
# For one class we need to define/pass same set of attibutes then we can setup as follows.
```

```python
class Emp:
    company = "AJA"
    def setup(obj, name, e_id): ## setup/adding data attributes here
        obj.name = name
        obj.id = e_id
    def get_details(obj):
        print(obj.name, obj.id)
jay = Emp()
Emp.setup(jay, "jairam",1)
vedha = Emp()
Emp.setup(vedha, "Vedhanvesh", 2)
Emp.get_details(jay)
Emp.get_details(vedha)
```

```python
# without setup can we call get_details?
class Emp:
    company = "AJA"
    def setup(obj, name, e_id): ## setup/adding data attributes here
        obj.name = name
        obj.id = e_id
    def get_details(obj):
        print(obj.name, obj.id)
jay = Emp()
Emp.setup(jay, "jairam",1)
vedha = Emp()
##Emp.setup(vedha, "Vedhanvesh", 2)
Emp.get_details(jay)
Emp.get_details(vedha)
```

```python
## so here we are giving a restriction i.e., call setup first then get_details, which means to set data we are calling separate method always(setup)

## To overcome that follow below: like setup data parallell to data objects define a class like below:
```

```python
class Emp:
    def __init__(obj, name, e_id): ## here we are giving 3 parameters
        print(obj, name, e_id)
jay = Emp("jairam",1) ## but here we are passing 2 attributes only
asee = Emp("asee",2) # internally calling __new__ method and then call data object
vedha = Emp("vedhanvesh",3) #type = Emp here., Python developers can define class on their own like Emp. Class basically used to define user define 
# datatypes. This is called abstraction.
```
