```python
#new_list = old_list.copy()
my_list = [1, 2, 3, 4]
new_list = my_list.copy()
print("Original list:", my_list)
print("Copied list:", new_list)

my_list = [1, 2, [3, 4]]
new_list = my_list.copy()
my_list[2][0] = 99
print("Original list:", my_list)
print("Copied list:", new_list)

import copy
my_list = [1, 2, [3, 4]]
new_list = copy.deepcopy(my_list)
my_list[2][0] = 99
print("Original list:", my_list)
print("Copied list (deepcopy):", new_list)
```

```python
from copy import deepcopy
k=[4000, 5000, 6000]
l=[1000,2000,3000,k,7000,8000,"python",(100,200,300)]
p=l
m=l.copy()
n=l[::]
q=deepcopy(l)
m[0]=10
print("id-l=", id(l))
print("id-p=", id(p))
print("id-m=", id(m))
print("id-n=", id(n))
print("id-q=", id(q))
print("id-l--1", id(l[-1]))
print("id-m--1", id(m[-1]))
print("id-n--1", id(n[-1]))
print("id-q--1", id(q[-1]))
```

```python
from copy import deepcopy
k=[4000, 5000, 6000]
l=[1000,2000,3000,k,7000,8000,"python",(100,200,300)]
p=l
m=l.copy()
n=l[::]
q=deepcopy(l)
m[0]=10
print("id-l=", id(l))
print("id-p=", id(p))
print("id-m=", id(m))
print("id-n=", id(n))
print("id-q=", id(q))
print("id-l--1", id(l[-1]))
print("id-m--1", id(m[-1]))
print("id-n--1", id(n[-1]))
print("id-q--1", id(q[-1]))

print("id-l-3", id(l[3]))
print("id-m-3", id(m[3]))
print("id-n-3", id(n[3]))
print("id-q-3", id(q[3]))
```
